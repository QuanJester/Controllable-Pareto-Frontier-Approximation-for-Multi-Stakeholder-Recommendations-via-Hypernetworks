# 02. Trực Quan Hóa Đường Biên Pareto & Chỉ Số Hypervolume

Notebook này hỗ trợ vẽ đồ thị đường biên Pareto (Pareto Frontier) giữa các mục tiêu:
- **Mục tiêu 1**: NDCG@10 / Recall@10 (Độ chính xác)
- **Mục tiêu 2**: Intra-List Diversity (Độ đa dạng)
- **Mục tiêu 3**: Provider Fairness / Novelty (Công bằng nhà cung cấp)

Đồng thời so sánh giữa phương pháp **Pareto Hypernetworks** và các **Baselines (Scalarization, MGDA)**.

```python
import numpy as np
import matplotlib.pyplot as plt

# Dữ liệu minh họa tập nghiệm Pareto
f1 = np.linspace(0.1, 0.9, 30)
# Mặt cong Pareto lý tưởng: f2 = sqrt(1 - f1^2)
f2_hypernet = np.sqrt(np.maximum(0, 1 - f1**2)) + np.random.normal(0, 0.01, size=30)

# Baselines điểm rời rạc
f1_baselines = np.array([0.2, 0.45, 0.7, 0.85])
f2_baselines = np.sqrt(1 - f1_baselines**2) - 0.05

plt.figure(figsize=(8, 6))
plt.plot(f1, f2_hypernet, 'o-', color='crimson', label='Pareto Hypernetwork (Continuous)', linewidth=2)
plt.scatter(f1_baselines, f2_baselines, color='navy', marker='^', s=100, label='Linear Scalarization Baselines')

plt.title('So sánh Đường biên Pareto (Accuracy vs. Diversity)', fontsize=14, fontweight='bold')
plt.xlabel('Diversity Metric (IL-Div)', fontsize=12)
plt.ylabel('Accuracy Metric (NDCG@10)', fontsize=12)
plt.legend(frameon=True, fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
