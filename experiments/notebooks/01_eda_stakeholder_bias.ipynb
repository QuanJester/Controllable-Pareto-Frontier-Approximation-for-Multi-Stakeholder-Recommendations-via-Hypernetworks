# 01. Phân Tích Dữ Liệu Khám Phá & Thiên Lệch Các Bên Liên Quan (EDA & Stakeholder Bias)

Notebook này phân tích sự mất cân bằng trong hệ thống gợi ý đa mục tiêu:
1. **Người dùng (Users)**: Phân phối độ dài tương tác, long-tail users.
2. **Nhà cung cấp / Vật phẩm (Items / Providers)**: Mức độ hiển thị của các sản phẩm phổ biến vs sản phẩm niche (Popularity bias & Fairness).
3. **Nền tảng (Platform)**: Cân bằng doanh thu, độ đa dạng chủ đề và độ phủ (Diversity & Coverage).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
%matplotlib inline
```

### 1. Giả lập Dữ liệu Tương tác (Synthetic Interactions)

```python
# Tạo phân phối Zipfian giả lập tương tác người dùng - vật phẩm
np.random.seed(42)
n_items = 1000
item_popularity = np.random.zipf(a=1.6, size=50000)
item_popularity = item_popularity[item_popularity < n_items]

plt.figure(figsize=(9, 4))
plt.hist(item_popularity, bins=50, color='teal', edgecolor='black', alpha=0.7)
plt.title('Phân phối mức độ phổ biến của sản phẩm (Long-tail Distribution)', fontsize=12)
plt.xlabel('Item ID (Sắp xếp theo độ phổ biến)')
plt.ylabel('Tần suất tương tác')
plt.yscale('log')
plt.show()
```

### 2. Đánh giá Hệ số Bất bình đẳng Gini (Gini Coefficient for Fairness)

```python
def gini_index(array):
    array = np.sort(np.asarray(array, dtype=np.float64))
    index = np.arange(1, array.shape[0] + 1)
    n = array.shape[0]
    return (np.sum((2 * index - n - 1) * array)) / (n * np.sum(array))

counts = pd.Series(item_popularity).value_counts().values
print(f"Hệ số Gini đo lường mức độ thiên lệch hiển thị: {gini_index(counts):.4f}")
```
